```python

import pandas as pd

# Load the four datasets
orders = pd.read_csv("orders.csv", encoding="latin-1")
order_details = pd.read_csv("order_details.csv", encoding="latin-1")
pizzas = pd.read_csv("pizzas.csv", encoding="latin-1")
pizza_types = pd.read_csv("pizza_types.csv", encoding="latin-1")

print("All four datasets loaded successfully!")
```

```python

print("Orders:", orders.shape)
print("Order Details:", order_details.shape)
print("Pizzas:", pizzas.shape)
print("Pizza Types:", pizza_types.shape)
```

```python

print("===== ORDERS =====")
print(orders.info())

print("\n===== ORDER DETAILS =====")
print(order_details.info())

print("\n===== PIZZAS =====")
print(pizzas.info())

print("\n===== PIZZA TYPES =====")
print(pizza_types.info())
```

```python

# Display the first 5 rows of each dataset

print("===== ORDERS =====")
display(orders.head())

print("===== ORDER DETAILS =====")
display(order_details.head())

print("===== PIZZAS =====")
display(pizzas.head())

print("===== PIZZA TYPES =====")
display(pizza_types.head())
```

```python

print("Pizza sizes:")
print(pizzas["size"].unique())

print("\nPizza categories:")
print(pizza_types["category"].unique())

print("\nOrder date range:")
print(orders["date"].min(), "to", orders["date"].max())
```

```python
# ==========================================
# STEP 2: DATA QUALITY CHECK
# ==========================================

# 1. Check missing values
print("===== MISSING VALUES =====")

print("\nOrders:")
print(orders.isnull().sum())

print("\nOrder Details:")
print(order_details.isnull().sum())

print("\nPizzas:")
print(pizzas.isnull().sum())

print("\nPizza Types:")
print(pizza_types.isnull().sum())


# 2. Check duplicate records
print("\n===== DUPLICATE RECORDS =====")

print("Orders duplicates:", orders.duplicated().sum())
print("Order Details duplicates:", order_details.duplicated().sum())
print("Pizzas duplicates:", pizzas.duplicated().sum())
print("Pizza Types duplicates:", pizza_types.duplicated().sum())


# 3. Check unique IDs
print("\n===== UNIQUE ID CHECK =====")

print("Unique order IDs:", orders["order_id"].nunique())
print("Total order records:", len(orders))

print("Unique pizza IDs:", pizzas["pizza_id"].nunique())
print("Total pizza records:", len(pizzas))


# 4. Check quantities and prices
print("\n===== VALUE CHECK =====")

print("Minimum quantity:", order_details["quantity"].min())
print("Maximum quantity:", order_details["quantity"].max())

print("Minimum price:", pizzas["price"].min())
print("Maximum price:", pizzas["price"].max())


# 5. Check data types
print("\n===== DATA TYPES =====")

print(orders.dtypes)
print(order_details.dtypes)
print(pizzas.dtypes)
print(pizza_types.dtypes)
```

```python
# ==========================================
# STEP 3: DATE AND TIME PREPARATION
# ==========================================

# Convert date column to datetime
orders["date"] = pd.to_datetime(orders["date"])

# Create useful date features
orders["year"] = orders["date"].dt.year
orders["month"] = orders["date"].dt.month
orders["month_name"] = orders["date"].dt.month_name()
orders["day_name"] = orders["date"].dt.day_name()

# Display updated data types
print("===== UPDATED DATA TYPES =====")
print(orders.dtypes)

# Display the first five records
print("\n===== UPDATED ORDERS =====")
display(orders.head())

# Check the date range
print("\n===== DATE RANGE =====")
print("Start date:", orders["date"].min())
print("End date:", orders["date"].max())
```

```python
# ==========================================
# STEP 4: COMBINE THE DATASETS
# ==========================================

# 1. Merge orders with order details
sales = orders.merge(
    order_details,
    on="order_id",
    how="inner",
    validate="one_to_many"
)

# 2. Merge with pizza information
sales = sales.merge(
    pizzas,
    on="pizza_id",
    how="left",
    validate="many_to_one"
)

# 3. Merge with pizza type information
sales = sales.merge(
    pizza_types,
    on="pizza_type_id",
    how="left",
    validate="many_to_one"
)

# Display the shape of the combined dataset
print("Combined dataset shape:", sales.shape)

# Display the columns
print("\n===== COMBINED COLUMNS =====")
print(sales.columns.tolist())

# Display the first five rows
print("\n===== COMBINED DATASET =====")
display(sales.head())
```

```python
# ==========================================
# STEP 4.3: CALCULATE REVENUE
# ==========================================

# Calculate revenue for each ordered pizza
sales["revenue"] = sales["quantity"] * sales["price"]

# Display important columns
display(
    sales[
        [
            "order_id",
            "date",
            "pizza_id",
            "name",
            "category",
            "size",
            "quantity",
            "price",
            "revenue"
        ]
    ].head(10)
)

# Check total revenue
print("Total revenue:", sales["revenue"].sum())

# Check total quantity sold
print("Total pizzas sold:", sales["quantity"].sum())
```

```python
# ==========================================
# STEP 5: VALIDATE THE COMBINED DATASET
# ==========================================

# 1. Check missing values after merging
print("===== MISSING VALUES AFTER MERGE =====")
print(sales.isnull().sum())


# 2. Check the shape after revenue calculation
print("\n===== DATASET SHAPE =====")
print("Rows:", sales.shape[0])
print("Columns:", sales.shape[1])


# 3. Check revenue values
print("\n===== REVENUE VALIDATION =====")
print("Minimum revenue:", sales["revenue"].min())
print("Maximum revenue:", sales["revenue"].max())
print("Total revenue:", sales["revenue"].sum())


# 4. Verify revenue calculation
calculated_revenue = sales["quantity"] * sales["price"]

print("\n===== REVENUE CALCULATION CHECK =====")
print(
    "Revenue calculation correct:",
    sales["revenue"].equals(calculated_revenue)
)


# 5. Check for invalid values
print("\n===== INVALID VALUE CHECK =====")
print("Quantity <= 0:", (sales["quantity"] <= 0).sum())
print("Price <= 0:", (sales["price"] <= 0).sum())
print("Revenue <= 0:", (sales["revenue"] <= 0).sum())
```

```python
# ==========================================
# STEP 6: MONTHLY SALES PERFORMANCE
# ==========================================

# Group sales by month
monthly_sales = (
    sales.groupby(["month", "month_name"])
    .agg(
        total_revenue=("revenue", "sum"),
        total_pizzas_sold=("quantity", "sum"),
        total_orders=("order_id", "nunique")
    )
    .reset_index()
    .sort_values("month")
)

# Calculate average order value
monthly_sales["average_order_value"] = (
    monthly_sales["total_revenue"] /
    monthly_sales["total_orders"]
)

# Display the monthly summary
print("===== MONTHLY SALES SUMMARY =====")

display(monthly_sales)
```

```python
# ==========================================
# STEP 6.2: MONTHLY REVENUE VISUALIZATION
# ==========================================

import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

plt.bar(
    monthly_sales["month_name"],
    monthly_sales["total_revenue"]
)

plt.title("Monthly Revenue Performance")
plt.xlabel("Month")
plt.ylabel("Revenue ($)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# STEP 6.3: MONTHLY PIZZA SALES VISUALIZATION
# ==========================================

plt.figure(figsize=(12, 6))

plt.bar(
    monthly_sales["month_name"],
    monthly_sales["total_pizzas_sold"]
)

plt.title("Monthly Pizza Quantity Sold")
plt.xlabel("Month")
plt.ylabel("Pizzas Sold")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# ===== PRODUCT PERFORMANCE ANALYSIS =====

product_performance = sales.groupby(
    ['pizza_id', 'name', 'category']
).agg(
    total_quantity_sold=('quantity', 'sum'),
    total_revenue=('revenue', 'sum'),
    average_price=('price', 'mean')
).reset_index()

# Sort products by revenue
product_performance = product_performance.sort_values(
    by='total_revenue',
    ascending=False
)

print("===== TOP 10 PIZZAS BY REVENUE =====")

display(product_performance.head(10))
```

```python
# ===== CATEGORY PERFORMANCE ANALYSIS =====

category_performance = sales.groupby(
    'category'
).agg(
    total_quantity_sold=('quantity', 'sum'),
    total_revenue=('revenue', 'sum'),
    number_of_products=('pizza_id', 'nunique')
).reset_index()

# Sort by total revenue
category_performance = category_performance.sort_values(
    by='total_revenue',
    ascending=False
)

print("===== CATEGORY PERFORMANCE =====")

display(category_performance)
```

```python
# ===== REVENUE BY PIZZA CATEGORY =====

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.bar(
    category_performance['category'],
    category_performance['total_revenue']
)

plt.title('Total Revenue by Pizza Category')
plt.xlabel('Pizza Category')
plt.ylabel('Total Revenue ($)')

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# ===== CATEGORY REVENUE CONTRIBUTION =====

total_revenue = category_performance['total_revenue'].sum()

category_performance['revenue_percentage'] = (
    category_performance['total_revenue'] / total_revenue
) * 100

display(
    category_performance[
        [
            'category',
            'total_revenue',
            'revenue_percentage'
        ]
    ]
)
```

```python
# ===== TOP 10 PIZZAS BY QUANTITY SOLD =====

top_quantity_products = sales.groupby(
    ['pizza_id', 'name', 'category']
).agg(
    total_quantity_sold=('quantity', 'sum'),
    total_revenue=('revenue', 'sum'),
    average_price=('price', 'mean')
).reset_index()

# Sort by quantity sold
top_quantity_products = top_quantity_products.sort_values(
    by='total_quantity_sold',
    ascending=False
)

print("===== TOP 10 PIZZAS BY QUANTITY SOLD =====")

display(top_quantity_products.head(10))
```

```python

# STEP 6A: Sales performance by hour

# Extract hour from time column
sales['hour'] = pd.to_datetime(
    sales['time'].astype(str),
    format='%H:%M:%S'
).dt.hour

# Group sales by hour
hourly_sales = sales.groupby('hour').agg(
    total_revenue=('revenue', 'sum'),
    total_pizzas_sold=('quantity', 'sum'),
    total_orders=('order_id', 'nunique')
).reset_index()

# Calculate average order value
hourly_sales['average_order_value'] = (
    hourly_sales['total_revenue'] / hourly_sales['total_orders']
)

# Display results
display(hourly_sales)

# Plot hourly revenue
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))
plt.bar(hourly_sales['hour'], hourly_sales['total_revenue'])

plt.title('Hourly Revenue Performance')
plt.xlabel('Hour of Day')
plt.ylabel('Revenue ($)')
plt.xticks(hourly_sales['hour'])
plt.grid(axis='y', alpha=0.3)

plt.show()
```

```python
# ==========================================
# SALES PERFORMANCE BY DAY OF THE WEEK
# ==========================================

# Group sales by day name
weekday_sales = sales.groupby(
    ['day_name']
).agg(
    total_revenue=('revenue', 'sum'),
    total_pizzas_sold=('quantity', 'sum'),
    total_orders=('order_id', 'nunique')
).reset_index()

# Define the correct weekday order
weekday_order = [
    'Monday',
    'Tuesday',
    'Wednesday',
    'Thursday',
    'Friday',
    'Saturday',
    'Sunday'
]

# Set weekday order
weekday_sales['day_name'] = pd.Categorical(
    weekday_sales['day_name'],
    categories=weekday_order,
    ordered=True
)

# Sort by weekday
weekday_sales = weekday_sales.sort_values('day_name')

# Calculate average order value
weekday_sales['average_order_value'] = (
    weekday_sales['total_revenue'] /
    weekday_sales['total_orders']
)

# Display results
print("===== WEEKDAY SALES PERFORMANCE =====")
display(weekday_sales)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.bar(
    weekday_sales['day_name'],
    weekday_sales['total_revenue']
)

plt.title('Revenue Performance by Day of the Week')
plt.xlabel('Day of the Week')
plt.ylabel('Total Revenue ($)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# PIZZA SIZE PERFORMANCE ANALYSIS
# ==========================================

size_performance = sales.groupby(
    ['size']
).agg(
    total_quantity_sold=('quantity', 'sum'),
    total_revenue=('revenue', 'sum'),
    total_orders=('order_id', 'nunique'),
    average_price=('price', 'mean')
).reset_index()

# Calculate average order value
size_performance['average_revenue_per_order'] = (
    size_performance['total_revenue'] /
    size_performance['total_orders']
)

# Sort by revenue
size_performance = size_performance.sort_values(
    'total_revenue',
    ascending=False
)

print("===== PIZZA SIZE PERFORMANCE =====")
display(size_performance)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.bar(
    size_performance['size'],
    size_performance['total_revenue']
)

plt.title('Revenue Performance by Pizza Size')
plt.xlabel('Pizza Size')
plt.ylabel('Total Revenue ($)')

plt.tight_layout()
plt.show()
```
